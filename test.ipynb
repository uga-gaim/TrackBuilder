```python
import pandas as pd
import track_builder as tb

# ================================================================
# 0) Parameters (adjust the path and year if needed)
# ================================================================
BASE_PATH = r"C:\Users\lamin\OneDrive\Documents\Maitrise en info Canada\ASTD project\data"
YEAR      = 2019
# At least 5 months (adjust as needed)
MONTHS_TO_LOAD = [5, 6, 7, 8, 9]

# Keep it lightweight: first and last day of each month + quality filter
USECOLS   = "default"
SAMPLING  = [0, -1]
QUAL_MIN  = 60


# ================================================================
# 1) Load the first and last days of at least 5 months
#    (respects the sampling setting sampling=[0, -1])
# ================================================================
df = tb.load_astd_monthly(
    BASE_PATH, YEAR, months=MONTHS_TO_LOAD, progress=True,
    usecols=USECOLS, sampling=SAMPLING, quality_threshold_minutes=QUAL_MIN
)
if df.empty:
    raise RuntimeError("Aucune donnée chargée. Vérifie chemin/année/mois.")
df["yyyymm"] = df["date_time_utc"].dt.strftime("%Y%m")

print("Mois couverts:", sorted(df["yyyymm"].unique().tolist()))
print(f"Total lignes (échantillon 1er/dernier jours/5+ mois): {len(df):,}")


# ================================================================
# 2) Build the track table (tracks) to associate segments
#    with tracks across months
#    -> based on the previous student's logic in main.py
# ================================================================
tracks = tb.build_track_table(df)

# Normalize a 'month_yyyymm' field for proper joining
if "month" in tracks.columns:
    # Generally 'YYYY-MM' -> 'YYYYMM'
    tracks = tracks.assign(month_yyyymm=tracks["month"].astype(str).str.replace("-", "", regex=False))
elif "date_time_utc" in tracks.columns:
    tracks = tracks.assign(month_yyyymm=pd.to_datetime(tracks["date_time_utc"], errors="coerce").dt.strftime("%Y%m"))
else:
    raise KeyError("La table tracks ne contient pas 'month' (ni date pour l’inférer).")

# Resolve identifier column names
pos_seg_col   = "shipid" if "shipid" in df.columns else ("segment_id" if "segment_id" in df.columns else None)  # côté positions
seg_col_track = "segment_id" if "segment_id" in tracks.columns else ("shipid" if "shipid" in tracks.columns else None)  # côté tracks
if pos_seg_col is None or seg_col_track is None:
    raise KeyError("Impossible de résoudre l’identifiant segment/navire (shipid/segment_id).")

print(f"Colonnes d’identifiants: positions={pos_seg_col}  tracks={seg_col_track}")


# ================================================================
# 3) Choose a track identifier (track_id) and find
#    all shipids (segments) for each month
#    -> select the track with the best coverage across months
# ================================================================
coverage = (
    tracks.groupby("track_id")["month_yyyymm"]
          .nunique()
          .sort_values(ascending=False)
)
if coverage.empty:
    raise ValueError("Aucun track_id trouvé dans la table de suivi.")

chosen_track = coverage.index[0]
covered_n    = int(coverage.iloc[0])
print(f"Track choisi: {chosen_track} (couvre {covered_n} mois)")

# shipid/segment_id per month for this track
per_month_shipids = (
    tracks.loc[tracks["track_id"] == chosen_track, ["month_yyyymm", seg_col_track]]
          .dropna()
          .groupby("month_yyyymm")[seg_col_track]
          .apply(lambda s: sorted(set(s.tolist())))
          .sort_index()
          .to_dict()
)
print("Ship IDs par mois (aperçu):")
for m, ids in per_month_shipids.items():
    print(f"  {m}: {len(ids)} id(s) → {ids[:5]}{'...' if len(ids)>5 else ''}")


# ================================================================
# 4) Load ship IDs for each month
#    (reload each month then filter by the shipids belonging to this track)
#    -> aligns with the approach used in the example
# ================================================================
def load_month_subset(year: int, month: int) -> pd.DataFrame:
    """Recharge le mois COMPLET (sans sampling) pour garantir des points présents pour le shipid."""
    return tb.load_astd_monthly(
        BASE_PATH, year, months=[month], progress=False,
        usecols=USECOLS, sampling=None,  # <<< IMPORTANT: sampling désactivé ici
        # quality_threshold_minutes=30  # tu peux activer si tu veux lisser
    )

positions_list = []
target_months  = sorted(set(tracks.loc[tracks["track_id"] == chosen_track, "month_yyyymm"]))

for m_str in target_months:
    y, mm = int(m_str[:4]), int(m_str[4:])
    dfi   = load_month_subset(y, mm)  # mois complet
    shipids = set(per_month_shipids.get(m_str, []))
    if not shipids:
        continue
    dfi = dfi[dfi[pos_seg_col].isin(shipids)].copy()
    positions_list.append(dfi)

positions_for_track = pd.concat(positions_list, ignore_index=True)
print(f"Positions retenues pour le track {chosen_track}: {len(positions_for_track):,} lignes")


# ================================================================
# 5) Plot the chosen track with its segments

# ================================================================
# a) Detailed view (follows the logic of the previous student)
fig_track = tb.plot_individual_track(
    chosen_track,
    tracks,
    positions_for_track,
    show_segments=True,                 # one polyline per segment (often = shipid)
    map_style="open-street-map",
    title=f"Track {chosen_track} – vue détaillée",
)
display(fig_track)


```

```python
# ===== 3’) Sélectionner un track couvrant EXACTEMENT les mois ET avec >=2 points par mois =====

target_set = {f"{YEAR}{m:02d}" for m in MONTHS_TO_LOAD}

# ensemble des mois par track_id (sur l’aperçu)
months_by_track = (tracks.groupby("track_id")["month_yyyymm"]
                   .agg(lambda s: set(s.dropna().astype(str).tolist())))

# candidats: ensemble == target_set
exact_tracks = months_by_track[months_by_track.apply(lambda s: s == target_set)]
if exact_tracks.empty:
    raise ValueError(f"Aucun track ne couvre EXACTEMENT {sorted(target_set)} (avec sampling={SAMPLING}).")

# ---- Charger chaque mois COMPLET une seule fois (cache), sans filtre qualité agressif ----
def load_month_full(year: int, month: int) -> pd.DataFrame:
    return tb.load_astd_monthly(
        BASE_PATH, year, months=[month], progress=False,
        usecols=USECOLS, sampling=None, quality_threshold_minutes=0
    )

month_cache = {}
for m_str in sorted(target_set):
    y, mm = int(m_str[:4]), int(m_str[4:])
    dfi = load_month_full(y, mm)
    if dfi.empty:
        month_cache[m_str] = pd.DataFrame(columns=df.columns)
    else:
        # on standardise le nom d’id côté positions
        if "shipid" in dfi.columns:
            dfi["_segpos"] = dfi["shipid"]
        elif "segment_id" in dfi.columns:
            dfi["_segpos"] = dfi["segment_id"]
        else:
            raise KeyError("Ni 'shipid' ni 'segment_id' dans les données rechargées.")
        month_cache[m_str] = dfi[["_segpos", "latitude", "longitude", "date_time_utc"]].copy()

# ---- Évaluer chaque candidat : faut >=2 points pour AU MOINS un shipid par mois ----
best_tid, best_total_pts = None, -1
best_per_month_ids = None

for tid in exact_tracks.index:
    # shipids annoncés par tracks pour chaque mois cible
    per_m_ids = (
        tracks.loc[(tracks["track_id"] == tid) & (tracks["month_yyyymm"].isin(target_set)), ["month_yyyymm", seg_col_track]]
              .dropna()
              .groupby("month_yyyymm")[seg_col_track]
              .apply(lambda s: sorted(set(s.tolist())))
              .to_dict()
    )

    ok = True
    total_pts = 0
    chosen_ids_per_month = {}

    for m_str in target_set:
        ids = per_m_ids.get(m_str, [])
        if not ids:
            ok = False; break

        dfi = month_cache[m_str]
        # nombre de points par shipid de ce mois
        counts = dfi["_segpos"].value_counts()

        # garder les shipid avec >=2 points
        good_ids = [sid for sid in ids if counts.get(sid, 0) >= 2]
        if not good_ids:
            ok = False; break

        # on peut en garder plusieurs ; on comptabilise
        month_pts = sum(int(counts.get(sid, 0)) for sid in good_ids)
        total_pts += month_pts
        chosen_ids_per_month[m_str] = good_ids

    if ok and total_pts > best_total_pts:
        best_tid, best_total_pts = tid, total_pts
        best_per_month_ids = chosen_ids_per_month

if best_tid is None:
    raise ValueError(
        "Parmi les tracks exacts, aucun n’a >=2 points pour chaque mois après rechargement complet. "
        "Relaxe la contrainte (par ex. QUAL_MIN=0, sampling=None pour la construction des tracks) "
        "ou essaie d’autres mois."
    )

chosen_track = best_tid
print(f"Track choisi (exact mois + >=2 pts/mois): {chosen_track} ({best_total_pts} points total)")

# ===== 4’) Construire positions_for_track à partir des shipid retenus (>=2 pts/mois) =====
positions_list = []
for m_str, ids in best_per_month_ids.items():
    y, mm = int(m_str[:4]), int(m_str[4:])
    dfi = month_cache[m_str]
    dfi = dfi[dfi["_segpos"].isin(set(ids))].copy()
    # remettre le bon nom de colonne pour la suite (plot)
    dfi[pos_seg_col] = dfi["_segpos"]
    positions_list.append(dfi)

positions_for_track = pd.concat(positions_list, ignore_index=True)
print(f"Positions retenues (>=2 pts/mois): {len(positions_for_track):,} lignes")

# ===== 5) Tracé identique =====
fig = tb.plot_individual_track(
    chosen_track, tracks, positions_for_track,
    show_segments=True,
    map_style="open-street-map",
    title=f"Track {chosen_track} – EXACT {sorted(target_set)} (>=2 pts/mois)",
)
fig.show()
```
